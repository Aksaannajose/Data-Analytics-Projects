```python
States_Stocks = df.groupby('state')['stocks'].sum().reset_index()
Top_5_Stocks_States = States_Stocks.sort_values(by='stocks', ascending=False).head(5)
Top_5_Stocks_States
```

```python
plt.figure(figsize=(12, 4))
sns.barplot(x='state', y='stocks', data=Top_5_Stocks_States, palette='viridis')
#plt.title('Top 5 States by Total Stocks')
plt.xlabel('State')
plt.ylabel('Total Stocks')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
honey_stock_states = df.groupby ('state')['stocks'].sum().reset_index()
Top_5_honey_stock_states = honey_stock_states.sort_values(by='stocks',ascending = False).reset_index().head(5)
Top_5_honey_stock_states.drop('index', axis =1, inplace = True)
Top_5_honey_stock_states

sns.barplot(data = Top_5_honey_stock_states, x = "state", y= "stocks")
#plt.title("Top_5_honey_stock_states")
plt.xlabel("state")
plt.ylabel("stocks")
```
