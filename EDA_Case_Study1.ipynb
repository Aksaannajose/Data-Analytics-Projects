```python
df = df.dropna(subset=['product']) # dropna - drop the null values , drop the rows which are having null values
```

```python
df.isnull().sum()
```

```python
# brand column as well we are having 1 null value we can either delete it or we can fill it with some specific value as well

df['brand'] = df['brand'].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
# replace the null value of description with value of your choice

df['description']  = df['description'].fillna("Description not available")
```

```python
df.isnull().sum()
```

```python
df['rating'] = df.groupby(['brand','category','sub_category'])['rating'].transform(lambda x: x.fillna(x.mean))
```
