#It was observed that the average purchase made by the men of the age 18-25 was 10000. Is it still the same?

```python
# write down null and alternate hypothesis.
# create a sub data based on condition
# we need to create a sample
# we perform hypothesis testing
# we conclude if it is null or alternate hypothesis
```

```python
# null - average purchase made by men of the age 18-25 age is 10000.
# alternate - average purchase made by men of the age 18-25 is not 100000.
```

```python
new_data = data.loc[(data['Age']==1)&(data['Gender']==1)]
```

```python
new_data
```

```python
sample_size=1000
```

```python
data.sample(5, random_state=0)
```

```python
sample=new_data.sample(1000,random_state=4)
```

```python
sample
```

```python
# one sample t test

from scipy.stats import ttest_1samp
```

```python
t_test,p_value=ttest_1samp(sample['Purchase'],10000)
```

```python
p_value
```

```python
if p_value<0.05:
  print('alternate - average purchase made by men of the age 18-25 is not 10000')
else:
  print('null - average purchase made by men of the age 18-25 is not 10000.')
```

#Is the average purchase made by men and women of the age 18-25 same?

```python
# write down null and alternate hypothesis.
# create a sub data based on condition
# we need to create sample
# we perform hypothesis testing
# we conclude if it is null or alternate hypothesis
```

```python
# null - average purchase made by men and women of the age 18-25 same
# alternate - average purchase made by men and women of the age 18-25 are not same
```

```python
data_men = data.loc[(data['Age']==1)&(data['Gender']==1)]
data_women = data.loc[(data['Age']==0)&(data['Gender']==1)]
```

```python
data_women
```

```python
data_men
```

```python
data_men.shape
```

```python
data_women.shape
```

```python
sample_size = 1000
```

```python
data_men_sample=data_men.sample(sample_size,random_state=0)
data_women_sample=data_women.sample(sample_size,random_state=0)
```

```python
data_men_sample
```

```python
data_women_sample
```

```python
print(data_men_sample['Purchase'].var())
print(data_women_sample['Purchase'].var())
```

```python
from scipy.stats import ttest_ind
t_stats,p_val=ttest_ind(data_men_sample['Purchase'],data_women_sample['Purchase'],equal_var=True)
```

```python
p_val
```

```python
if p_value<0.05:
  print('Alternate')
else:
  print('Null')
```
