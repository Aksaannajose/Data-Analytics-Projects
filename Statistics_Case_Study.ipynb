```python
# find the mean, median and variance, standard deviation, range of population in thousand
```

```python
mean_population = df['Population in thousands (2017)'].mean()
median_population = df['Population in thousands (2017)'].median()
variance = df['Population in thousands (2017)'].var()
stddev = df['Population in thousands (2017)'].std()
range = df['Population in thousands (2017)'].max() - df['Population in thousands (2017)'].min()

print("Mean Population",mean_population)
print("Median Population",median_population)
print("Variance",variance)
print("Standard Deviation",stddev)
print("Range",range)
```

```python
df['GDP per capita (current US$)'].skew()
```

```python
gdp_skewness = df['GDP: Gross domestic product (million current US$)'].skew()
print(f"Skewness of GDP: {gdp_skewness:.2f}")

if gdp_skewness > 0.5:
  skew_type = "Positively (right) skewed"
elif gdp_skewness < -0.5:
  skew_type = "negatively (left) skewed"
else:
    skew_type = "approximatelly symmetrical"

print(f"The GDP distribution is {skew_type}.")
```
