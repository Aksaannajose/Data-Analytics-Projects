# 2 The Average monthly hours of a employee having 2 years of experice is 167. Is't it same for the employee having more then 2 years of experice?


```python
new_df=df1=df[(df.satisfactoryLevel>=0) & (df.satisfactoryLevel<=0.50) & (df['timeSpent.company']>2)&( df.avgMonthlyHours>=130)& (df.avgMonthlyHours<=200)]
```

```python
new_df
```

```python
proposed_mean=167
```

```python
sample_df=new_df.sample(100,random_state=0)
```

```python
sample_df
```

```python
#t-test
from scipy.stats import ttest_1samp
```

```python
stats,p_value=ttest_1samp(sample_df['avgMonthlyHours'],proposed_mean)
```

```python
print(p_value)
```

```python
alpha = 0.05
if p_value<alpha:
    print('There is significant difference between average monthly working hours')
else:
    print('There is no significant difference in average monthly working hours')
```
